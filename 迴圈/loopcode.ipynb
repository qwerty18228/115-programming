```python
for j in range(1,6):
  print(f'1*{j}={1*j}',end=',')
print()

for j in range(1,6):
  print(f'2*{j}={2*j}',end=',')
print()

for j in range(1,6):
  print(f'3*{j}={3*j}',end=',')
print()
```

```python
for i in range(1,6):
    for j in range(1,6):
      print(f'{i}*{j}={i*j}',end=',')
    print()
```
